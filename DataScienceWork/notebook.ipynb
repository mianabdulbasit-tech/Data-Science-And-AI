```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig = go.Figure()

fig.add_trace(go.Scatter(x=x, y=y))

fig.update_layout(
title='Interactive Plot with Plotly',
xaxis_title='X-axis',
yaxis_title='Y-axis',
font=dict(
family="Courier New, monospace",
size=18,
color="RebeccaPurple"
)
)

fig.show()
```
