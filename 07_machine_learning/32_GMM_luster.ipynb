# Gaussian Mixture Model (GMM) Clustering

```python
# Import required libraries
import numpy as np
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs

# Generate a sample dataset
X, y_true = make_blobs(n_samples=400, centers=4, cluster_std=0.60, random_state=0)

# Create the Gaussian Mixture Model
gmm = GaussianMixture(n_components=4, random_state=42)

# Train (fit) the model on the dataset
gmm.fit(X)

# Predict the cluster label for each data point
labels = gmm.predict(X)

# Display the first 10 predicted cluster labels
print(labels[:10])
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn import datasets

#  load the Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Use only the first two features for visualization

# Create the Gaussian Mixture Model with 3 components (clusters)
gmm = GaussianMixture(n_components=3, random_state=42)

# Train (fit) the model on the dataset
gmm.fit(X)

# Predict the cluster label for each data point
labels = gmm.predict(X)

# visualize the clustering results
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.title('GMM Clustering on Iris Dataset')
plt.show()
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn import datasets

#  load the Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Use only the first two features for visualization

# Create the Gaussian Mixture Model with 3 components (clusters)
gmm = GaussianMixture(n_components=3, random_state=42)

# Train (fit) the model on the dataset
gmm.fit(X)

# Predict the cluster label for each data point
labels = gmm.predict(X)

# compute the log-likelihood of the data under the model
log_likelihood = gmm.score(X)

# print the log-likelihood
print("log-likelihood:", log_likelihood)

# visualize the clustering results
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.title('GMM Clustering on Iris Dataset')
plt.show()
```

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn import datasets

#  load the Iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Use only the first two features for visualization

# Create the Gaussian Mixture Model with 3 components (clusters)
gmm = GaussianMixture(n_components=3, random_state=42)

# Train (fit) the model on the dataset
gmm.fit(X)

# Predict the cluster label for each data point
labels = gmm.predict(X)

# compute the log-likelihood of the data under the model
log_likelihood = gmm.score(X)

# compute the AIC and BIC for the model
aic = gmm.aic(X) # AIC stands for (Akaike Information Criterion)
bic = gmm.bic(X) # BIC stands for (Bayesian Information Criterion)

# print the log-likelihood
print("log-likelihood:", log_likelihood)
print("AIC:", aic)
print("BIC:", bic)

# visualize the clustering results
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.title('GMM Clustering on Iris Dataset')
plt.show()
```
