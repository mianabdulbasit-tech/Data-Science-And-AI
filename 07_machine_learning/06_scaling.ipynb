# 1. Standard Scaling

`Standard scaling` is a technique used in data preprocessing to normalize the features of a dataset. It transforms the data so that it has a mean of 0 and a standard deviation of 1. This is particularly useful for algorithms that are sensitive to the scale of the data, such as gradient descent-based methods.

the formula is as follows:

z = (x - μ) / σ
Where:
- z is the standardized value
- x is the original value
- μ is the mean of the feature
- σ is the standard deviation of the feature


```python
# import libraries
#import pandas as pd 
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler
```

```python
# make an example dataset
df = {
    'age': [25, 30, 35, 40, 45],
    'height': [165, 170, 175, 180, 185],
    'weight': [55, 60, 65, 70, 75]
}

# convert the dataset into a pandas DataFrame
df = pd.DataFrame(df)
df.head()
```

```python
# import the scaler 
scaler = StandardScaler()

# fit the scaler to the data
scaled_df = scaler.fit_transform(df)
scaled_df
```

```python
# convert this data into a pandas DataFrame
scaled_df = pd.DataFrame(scaled_df, columns=df.columns)
scaled_df.head()
```

# Normalization


# L2 Normalization:

Remember that L2 normalization scales the data in such a way that the sum of the squares of the values in each row equals 1. This is particularly useful when you want to ensure that each feature contributes equally to the distance calculations in algorithms like k-nearest neighbors or support vector machines.

```python
from sklearn.preprocessing import Normalizer

data = [[ 1,  1,  1], [ 1,  1,  0], [ 1,  0,  0]]
normalizer = Normalizer(norm='l2')

print(normalizer.fit_transform(data))
```

L1 Normalization:
Also rescales each sample(row) but the different approach is that it scales the data in such a way that the sum of the absolute values of the values in each row equals 1. This can be useful when you want to ensure that each feature contributes equally to the distance calculations, but you want to be less sensitive to outliers compared to L2 normalization.

```python
from sklearn.preprocessing import Normalizer

data = [[ 1,  1,  1], [ 1,  1,  0], [ 1,  0,  0]]
normalizer = Normalizer(norm='l1')

print(normalizer.fit_transform(data))
```

```python
# generate non-normal data (exponential Distribution)
np.random.seed(0)
df = np.random.exponential(size = 1000 , scale=2)
df = pd.DataFrame(df, columns=['values'])
df.head()
```

```python
from sklearn.preprocessing import PowerTransformer
from sklearn.preprocessing import QuantileTransformer

pt_boxcox = PowerTransformer(method = 'box-cox')
pt_yeo_johnson = PowerTransformer(method= 'yeo-johnson')
qt_normal = QuantileTransformer(output_distribution = 'normal')

# boxcox k liye data must be positive
df['Box_cox'] = pt_boxcox.fit_transform(df[['values']] + 1)
df['Yeo_johnson'] = pt_yeo_johnson.fit_transform(df[['values']])
df['Quantile'] = qt_normal.fit_transform(df[['values']])
```

```python
df.head()
```

```python
# creat histograms for all columns using sns.hist and kde =True use for loop

for col in df.columns:
    sns.histplot(df[col], kde = True)
    plt.show()
```
