# Dat Merge

```python
import pandas as pd

# example datasets for data merging
# dataset 1:  Pakistani Individuals
data1 = {
    'id' : [1,2,3],
    'name' : ['Ahmed', 'Fatima', 'Ali'],
    'age' : [28, 32, 24]
}
df1 = pd.DataFrame(data1)
df1
```

```python
# Dataset 2: Additional information
data2 = {
    'id' : [1,2,4],
    'city' : ['Karachi', 'Lahore', 'Islamabad'],
    'occupation' : ['Engineer', 'Doctor', 'Teacher']
}
df2 = pd.DataFrame(data2)
df2
```

```python
# merging the dataframes on 'id'
merged_df = pd.merge(df1, df2, on= 'id' ,  how = 'inner')
merged_df
```

```python
# merging the dataframes on 'id'
merged_df = pd.merge(df1, df2, on= 'id' ,  how = 'outer')
merged_df
```

```python
# merging the dataframes on 'id'
merged_df = pd.merge(df1, df2, on= 'id' ,  how = 'left')
merged_df
```

```python
# merging the dataframes on 'id'
merged_df = pd.merge(df1, df2, on= 'id' ,  how = 'right')
merged_df
```

# Data Concatination

```python
import pandas as pd

# example datasets for Pakistani Context
# dataset 1:  Pakistani and ages
data1 = {
   
    'Name' : ['Ahmed', 'Fatima', 'Ali'],
    'Age' : [28, 32, 24]
}

# Data 2 : Pakistani and ages
data2 = {
   
    'Name' : ['Sara', 'Usman', 'Aisha'],
    'Age' : [29, 35, 27]
}

# Dataset 3: Additional information
data3 = {
   
    'City' : ['Karachi', 'Lahore', 'Islamabad'],
    'Province' : ['Sindh', 'Punjab', 'Islamabad']
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df3 = pd.DataFrame(data3)

# Concatenating the Rows
concatenated_rows = pd.concat([df1, df2], axis = 0)

#concatenating columns
# Note: For a meaningful column-wise  concatenation , dataframe should have the same
concatenated_columns = pd.concat([df1, df3],  axis=1)

print(concatenated_rows)

print("-------------------")

concatenated_columns
```
