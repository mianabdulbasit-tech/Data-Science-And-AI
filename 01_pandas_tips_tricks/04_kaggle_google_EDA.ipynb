# we needs to convert the KBs into MBs and then remove the Mbs

```python
import pandas as pd

def convert_size_to_mb(size):
    if pd.isna(size):
        return None
    size = size.strip()
    
    if size.endswith('M'):
        return float(size.replace('M', ''))
    elif size.endswith('k'):
        return float(size.replace('k', '')) / 1024
    else:
        return None   # for unexpected values

# apply function to column
df['Size_MB'] = df['Size'].apply(convert_size_to_mb)
```

#  Remove + sign from install columns and convert into  numeric values


```python
df['Installs_numeric'] = (
    df['Installs']
    .str.replace('+', '', regex=False)
    .str.replace(',', '')
    .astype(int)
)
```

# Now Remove the dollor($) sign in price columns 

```python
df['Price_numeric'] = (
    df['Price']
    .str.replace('$', '', regex=False)
    .astype(float)
)
```
