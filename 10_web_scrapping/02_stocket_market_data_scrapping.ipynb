```python
import yfinance as yf
import pandas as pd

# define the ticker symbol
tickerSymbol = 'AAPL' # AAPL keywords used for apple stocket data 

# get data on this ticker
tickerData = yf.Ticker(tickerSymbol)
# tickerData.info

# get the historical prices for this ticker
# `period` and `start/end` may conflict in some yfinance versions.
# Use a date range with an explicit interval instead.
tickerDf = tickerData.history(interval='1d', start='2010-1-1', end='2020-1-25')

# last closing price
tickerDf['Close'].iloc[-1]

tickerDf.head()
```

```python
# check the data shape
print(tickerDf.shape)
```

```python
import pandas as pd
import yfinance as yf
import datetime 
from datetime import date, timedelta

today = date.today()
d1 = today.strftime("%Y-%m-%d")
d1

# d2 = (today + timedelta(365)).strftime("%Y-%m-%d")
d2 = (today - timedelta(365)).strftime("%Y-%m-%d")
d2

start_date = d2
end_date = d1

# define the ticker symbol
tickerSymbol = 'TSLA' # TSLA keywords used for tesla data 

# get data on this ticker
tickerData = yf.Ticker(tickerSymbol)
# tickerData.info

# get the historical prices for this ticker
# `period` and `start/end` may conflict in some yfinance versions.
# Use a date range with an explicit interval instead.
tickerDf = tickerData.history(interval='1d', start=start_date, end=end_date)

tickerDf.head()

tickerDf.to_csv('TSLA.csv')
```
