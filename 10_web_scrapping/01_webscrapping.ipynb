```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://en.wikipedia.org/wiki/World_population"

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, "html.parser")

print(soup.title)
title = soup.title.get_text(strip=True)
print(title)

# print(soup.prettify()) #  print the whole html code
```

```python
# find all the tables 
tables = soup.find_all("table")

dataframe = [] # empty list

for idx, table in enumerate(tables):
    rows = table.find_all("tr")[1:]  # ignore the first row (header)
    data = []
    for row in rows:
        columns = row.find_all("td")
        cols = [col.get_text(strip=True) for col in columns]
        data.append(cols)
    df = pd.DataFrame(data)
    dataframe.append(df)

dataframe[1].to_csv("world_population.csv", index=False, header=True)
```
