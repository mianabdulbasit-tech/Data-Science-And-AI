```python
# import libraries
import pandas as pd
import numpy as np 
import seaborn as sns  
import plotly.express as px
```

```python
names = ['M Abdul Basit', 'Hamza', 'Arshad']
heights = [5.8, 5.9, 5.7]

# create a plot (line plot)
fig = px.line(x=names, y=heights, title='Heights of Individuals')

# show thw fig
fig.show()

# save the plot in phg file
fig.write_image('./outputs/01_lineplot.png')

# save as standalone html file
fig.write_html('./outputs/01_lineplot.html')
```

```python
names = ['M Abdul Basit', 'Hamza', 'Arshad']
heights = [5.8, 5.9, 5.7]

# create a plot (line plot)
fig = px.bar(x=names, y=heights, title='Heights of Individuals')

# show the figure
fig.show()

# save the plot in png file
fig.write_image('./outputs/01_barplot.png')

# save as standalone html file
fig.write_html('./outputs/01_barplot.html')
```

# Plots On titanic Dataset

```python
# import dataset
df = sns.load_dataset('titanic')
# check the head of dataset
df.head()
```

```python
# scatter plot

# if the nan values available in  the daraset then how i can remove the nan values from the dataset
# df.dropna(inplace=True)
fig  = px.scatter(df, x='age', y='fare', color='sex', size='fare', title='Plot of Age vs Fare of titanic Dataset')
fig.show()
```

```python
# scatter plot

# if the nan values available in  the daraset then how i can remove the nan values from the dataset
df.dropna(inplace=True)

fig  = px.scatter(df, x='age', y='fare', color='sex', size='age', title='Plot of Age vs Fare of titanic Dataset')
fig.show()

# save the plot in the png file
fig.write_image('./outputs/01_scatterplot.png', scale=3)

 # how to save in 300 dpi resolution
fig.write_image('./outputs/01_scatterplot_300dpi.png', scale=3, width=800, height=600)

# save file in the json format
fig.write_json('./outputs/01_scatterplot.json')
```

```python
import plotly.graph_objects as go

# create a simple line chart
fig = go.Figure(data=go.Scatter(x=[1, 2, 3, 4], y=[10, 11, 12, 13]))
fig.show()
```

```python
# custimize/Update the plot

fig.update_layout(
    title = 'Line Chart Plot Of Dummy Data',
    xaxis = dict(title='Numbers'),
    yaxis = dict(title='Values'),
    showlegend = True,
    #font=dict(family='Arial', size=14, color='blue')
)
fig.show()
```

```python
# heatmap in Plotly

data = np.random.rand(10, 10)
heatmap = go.Heatmap(z=data)
fig = go.Figure(data=heatmap)
fig.show()
```
