# Sunburst Plot To Show Heirarchy

```python
# Create sunburst plot on titanic dataset
import plotly.express as px
import seaborn as sns

# load titanic dataset
titanic = sns.load_dataset('titanic')

# create a sunburst plot
fig = px.sunburst(titanic, path=['sex', 'class', 'who', 'alive', 'alone'],
values='survived', color= 'sex')

# show the graph
fig.show()

```

```python
# Create sunburst plot on tips dataset
import plotly.express as px
import seaborn as sns

# create a sunburst plot
import plotly.express as px
import seaborn as sns

# load dataset
tips = sns.load_dataset("tips")

# convert categorical columns to string
tips['sex'] = tips['sex'].astype(str)
tips['smoker'] = tips['smoker'].astype(str)
tips['day'] = tips['day'].astype(str)
tips['time'] = tips['time'].astype(str)

# create sunburst
fig = px.sunburst(tips,
                     path=['sex', 'smoker', 'day', 'time'],
                     values='size', color='sex')

# how can we add more values when we hover the pointer on nthe plot
fig.update_traces(hovertemplate='<b>%{label}</b><br>Size: %{value}<br>Total Bill: %{customdata[0]}<br>Tip: %{customdata[1]}',
                  customdata=tips[['total_bill', 'tip']].values)
                  
# show the graph
fig.show()
```
