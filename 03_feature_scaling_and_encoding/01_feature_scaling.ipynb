## 1. "Min Max Scaling"

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Sample data
data = {'Value': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

df.head()
```

```python
# scale the data using min max scaler

# Min-Max Scaling
scaler = MinMaxScaler()
df['Scaled_Value'] = scaler.fit_transform(df['Value'].values.reshape(-1,1))
df.head()
```

```python

# Min-Max Scaling
scaler = MinMaxScaler()
df['Scaled_Value'] = scaler.fit_transform(df[['Value']])
print(df)
```

# 2. Standard Scaler or Z-score normalozation


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Sample data
data = {'Value': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)

df.head()
```

```python
# Stabdard Scaling
scaler = StandardScaler()
df['Scaled_Value'] = scaler.fit_transform(df[['Value']])
df.head()
```

# 3. Robust Scaler

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler

# Sample data  with outliers
data = {'Value': [10, 20, 30, 1000, 50]}
df = pd.DataFrame(data)

# Robust Scaling
scaler= RobustScaler()
df['Scaler_Value'] = scaler.fit_transform(df[['Value']])
print(df)
```

# 4. logrithmic Scaling /Normalization

```python
# import libraries
import numpy as np
import pandas as pd

# random data 
data = {'Value': [10000, 20000, 30000, 100000, 50000]}
df = pd.DataFrame(data)

# log Tranform
df['scaled_value'] =  np.log(df['Value'])
df['scaled_value10'] =  np.log10(df['Value'])
df['scaled_value2'] =  np.log2(df['Value'])

df.head()
```
