# Linear Algebra In Python 

```python
import numpy as np
```

Algebric equations:

2x + 3y = 8

5x + 4y = 20 

```python
# coefficient matrix A
A = np.array([[2,3], [5,4]]) # 2 by 2
A
```

```python
# constant matrix B
B = np.array([8, 20])
B
```

```python
# solve for x
x = np.linalg.solve(A,B) #solve for x
x
```

# `Solving Complex Equations`

\[
\begin{align*}
(1 + 2i)x + (2 -i)y & = 4 + i \\
(3 - i)x +4y &= 3 + 4i
\end{align*}
\]

```python
# Coefficients matrix A with complex numbers
A_complex = np.array([[1+2j, 2-1j], [3-1j, 4]])
A_complex
```

```python
# Constants matrix B with complex numbers
B_complex = np.array([4+1j, 3+4j])
B_complex
```

```python
# Solving for X in the complex case
X_complex = np.linalg.solve(A_complex, B_complex)
X_complex
```

```python
# Coefficients matrix A with complex numbers
A_complex = np.array([[1+2j, 2-1j], [3-1j, 4]])

# Constants matrix B with complex numbers
B_complex = np.array([4+1j, 3+4j])

# Solving for X in the complex case
X_complex = np.linalg.solve(A_complex, B_complex)

X_complex
```

# Linear Transformation

\[
\begin{bmatrix}
\cos(45^\circ) & -\sin(45^\circ) \\
\sin(45^\circ), & \cos(45^\circ)
\end{bmatrix} 
\]

```python
# create a vector  
v = np.array([3,3])
v
 
```

```python
# Define the rotation matrix for 45 degree rotation

theta = np.radians(45)  # Converting 45 degrees to radians 
rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)], 
                            [np.sin(theta), np.cos(theta)]])

print(rotation_matrix)
print("-------------------------")
print(v)
# apply the transformtion
v_transformed = np.dot(rotation_matrix, v)
v_transformed
```

```python
# create a vector  
v = np.array([3,3])
v
 # plot the vector v 
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original Vector')
plt.quiver(0, 0, v_transformed[0], v_transformed[1], angles='xy', scale_units='xy', scale=1, color='b', label='transformed Vector')
plt.xlim(-6, 6)
plt.ylim(-6, 6)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.axhline(y=0, color='k', linewidth= 0.5)
plt.axvline(x=0, color='k', linewidth= 0.5)
plt.title('2D Vector Transformation')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# Shear  Tranormation

```python
# Define the rotation matrix for 45 degree rotation

theta = np.radians(85)  # Converting 45 degrees to radians 
shear_matrix = np.array([[np.cos(theta), -np.sin(theta)], 
                            [np.sin(theta), np.cos(theta)]])

print(shear_matrix)
print("-------------------------")
print(v)
# apply the transformtion
v_sheared = np.dot(shear_matrix, v)
v_sheared
```

```python
# Define the shear matrix for shearing along the x-axis with k = 1.5
shear_factor = 1.5
shear_matrix = np.array([[1, shear_factor], [0, 1]])

# Applying the shear transformation
v_sheared =  np.dot(shear_matrix, v)

 # plot the original, rotated, and  sheared vectors
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original Vector')
plt.quiver(0, 0, v_transformed[0], v_transformed[1], angles='xy', scale_units='xy', scale=1, color='b', label='transformed Vector')
plt.quiver(0, 0, v_sheared[0], v_sheared[1], angles='xy', scale_units='xy', scale=1, color='g', label='Shear transformed Vector')
plt.xlim(-5, 10)
plt.ylim(-5, 10)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.axhline(y=0, color='k', linewidth= 0.5)
plt.axvline(x=0, color='k', linewidth= 0.5)
plt.title('2D Vector Transformation')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
# Adjusting the plot to include the effect of shearing on thee grid lines
# creating a grid of points
x = np.linspace(-5, 10, 10)
y = np.linspace(-5, 10, 10)
X, Y = np.meshgrid(x, y)

# Applying the shear transformation to the grid points
X_sheared, Y_sheared =np.dot(shear_matrix, [X.ravel(), Y.ravel()])
X_sheared = X_sheared.reshape(X.shape)
Y_sheared = Y_sheared.reshape(Y.shape)

v_sheared =  np.dot(shear_matrix, v)
 # plot the original, rotated, and  sheared vectors
plt.figure()
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='r', label='Original Vector')
plt.quiver(0, 0, v_transformed[0], v_transformed[1], angles='xy', scale_units='xy', scale=1, color='b', label='transformed Vector')
plt.quiver(0, 0, v_sheared[0], v_sheared[1], angles='xy', scale_units='xy', scale=1, color='g', label='Shear transformed Vector')

for i in range(len(x)):
    plt.plot(X_sheared[i, :], Y_sheared[i, :], 'k--', linewidth= 0.5)
    plt.plot(X_sheared[i, :], Y_sheared[i, :], 'k--', linewidth= 0.5)

plt.xlim(-5, 10)
plt.ylim(-5, 10)
# plt.grid(True, which='both', linestyle='--', linewidth=0.5)
# plt.axhline(y=0, color='k', linewidth= 0.5)
# plt.axvline(x=0, color='k', linewidth= 0.5)
plt.title('2D Vector Transformation With Sheared Grid')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# SVD In Python

```python
# Generate a matrix '4 by 3' dataset
A = np.random.rand(4, 3)
A
```

```python
# lets apply SVD
U, Sigma, VT = np.linalg.svd(A)

# print each matrix separately
print("U Matrix = ", U)
print("--------------------")
print("Sigma Matrix = ", Sigma)
print("--------------------")
print("VT Matrix = ", VT)
```
