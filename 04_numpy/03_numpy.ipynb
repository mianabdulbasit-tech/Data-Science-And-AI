# Enlist boolean operators conditional operators
| Operator | Meaning                  | Example  |
| -------- | ------------------------ | -------- |
| `==`     | Equal to                 | `a == b` |
| `!=`     | Not equal to             | `a != b` |
| `>`      | Greater than             | `a > b`  |
| `<`      | Less than                | `a < b`  |
| `>=`     | Greater than or equal to | `a >= b` |
| `<=`     | Less than or equal to    | `a <= b` |

# Making an array from a data or another array

```python
a = np.arange(0 , 222, 5)
a
```

```python
len(a)
```

```python
array1 = a[40 :45]
array1
```

```python
# stack the arrays

a1 = np.array([[1, 1],
               [2, 2]])

a2 = np.array([[3, 3],
               [4, 4]])
```

```python
# vertically stack the arrays
v = np.vstack((a1 , a2))
v
```

```python
# Horizontally stack the arrays
h = np.hstack((a1 , a2))
h
```

# # Split the arrays into 2D , 3D and upto so on


```python
arr = np.arange(1 , 25 )
arr
```

```python
# split the arrays into 2D array

np.split(arr, 2)
```

```python
# split the arrays
arr = np.arange(1 , 25 )
arr.reshape(2,12)
```

```python
# split the 2D array into three equal shape arrays

np.split(arr, 3)
```

```python
np.hsplit(arr, (3, 4))
```

```python
# create a matrix of big dimension
b = np.arange(0, 100, 2).reshape(5 , 10)
b
```

```python
# slicing a matrix 
b[:,:] # means the all the rows and ll the columns 
```

```python
# slicing a matrix 

c = b[1:3, 3:5] # 1:3 deine the eange of the matrix and 3:5 define the range the columns
```

```python
# sum the matrix all values
b.sum()
```

```python
b
```

```python
# check the maximum value from th matrix
b.max()
```

```python
# show the max row
b.max(axis=0)
```

```python
# show the highest columns
b.max(axis=1)
```

```python
b
```

```python
# check the minimum value from th matrix
b.min()
```

```python
# show the minimum complete row
b.min(axis=0)
```

```python
# show the minimum complete column

b.min(axis=1)
```
