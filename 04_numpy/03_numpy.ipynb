# Numpy Indexing and Slicing

```python
import numpy as np
```

```python
# create an array

a = np.array([1, 2, 3, 4, 5, 6, 7])
a
```

```python
# indexing

#a[1]  # 2
#a[-1]  # 7
# a[-2]  # 6
```

```python
# slicing
a[0:]
a[0: 5]
a[3: 6]
```

```python
b = np.arange(0, 12, 2)
b
```

```python
b[2]
```

```python
c = b.reshape(2,3)
c
```

```python
c[0]
```

```python
d = b.reshape(3,2)
d
```

```python
d[0]
```

```python
# multi conditions
c[c < 5 ]
```
