# Numpy Indexing and Slicing

```python
import numpy as np
```

```python
# create an array

a = np.array([1, 2, 3, 4, 5, 6, 7])
a
```

```python
# indexing

#a[1]  # 2
#a[-1]  # 7
# a[-2]  # 6
```

```python
# slicing
a[0:]
a[0: 5]
a[3: 6]
```

```python
b = np.arange(0, 12, 2)
b
```

```python
b[2]
```

```python
c = b.reshape(2,3)
c
```

```python
c[0]
```

```python
d = b.reshape(3,2)
d
```

```python
d[0]
```

```python
# multi conditions
c[c < 5 ]
```

# Enlist boolean operators conditional operators
| Operator | Meaning                  | Example  |
| -------- | ------------------------ | -------- |
| `==`     | Equal to                 | `a == b` |
| `!=`     | Not equal to             | `a != b` |
| `>`      | Greater than             | `a > b`  |
| `<`      | Less than                | `a < b`  |
| `>=`     | Greater than or equal to | `a >= b` |
| `<=`     | Less than or equal to    | `a <= b` |

# Making an array from a data or another array

```python
a = np.arange(0 , 222, 5)
a
```

```python
len(a)
```

```python
array1 = a[40 :45]
array1
```

```python
# stack the arrays

a1 = np.array([[1, 1],
               [2, 2]])

a2 = np.array([[3, 3],
               [4, 4]])
```

```python
# vertically stack the arrays
v = np.vstack((a1 , a2))
v
```

```python
# Horizontally stack the arrays
h = np.hstack((a1 , a2))
h
```

# # Split the arrays into 2D , 3D and upto so on


```python
arr = np.arange(1 , 25 )
arr
```

```python
# split the arrays into 2D array

np.split(arr, 2)
```

```python
# split the arrays
arr = np.arange(1 , 25 )
arr.reshape(2,12)
```

```python
# split the 2D array into three equal shape arrays

np.split(arr, 3)
```

```python
np.hsplit(arr, (3, 4))
```
