### Automatic EDA using `ydata_profiling`

```python
import pandas as pd
import seaborn as sns
import ydata_profiling as yd
```

```python
# import data from seaborn 
df = sns.load_dataset('titanic')
```

```python
# 1-ydata profile report (in same folder)
profile = yd.ProfileReport(df)
profile.to_file(output_file='ydata_titanic.html') # To save as an HTML file
```

```python
# 2-ydata profile report (inside the other folder)
profile = yd.ProfileReport(df)
profile.to_file(output_file='./outputs/03_ydata_titanic.html') # To save as an HTML file
```
