# Interpretation:

1. We have age of the passengers at x-axis and fare at y-axis

# Change the box plot into boxen plot

```python
# Boxen plot

# st=le of the plot
sns.set_theme(style='darkgrid')

# Boxen plot size of figure size
plt.figure(figsize=(12,6))
# change the scatter point colors 
pallete = {'First': 'green', 'Second': 'red', 'Third': 'blue'} # pallete mean color scheme / color mapping / color counter
sns.boxenplot(data= df, x= 'sex', y= 'age', hue='class', palette=pallete)

# laabels and title

# labels 
xlab = 'Gender'
ylab = 'Age'
plt.xlabel(xlab)
plt.ylabel(ylab)

# title
plt.title('BOX PLOT Of Age (Based On Gender) And Class Name Of Passengers')

# if we save to the file in other folder 

# saving a plot 300dpi, 600dpi into png file and size of the plot 
plt.savefig('./outputs/plots/Boxen_plot_zz.png', dpi=300)

# saving a plot in vector file such as svg file
plt.savefig('./outputs/plots/Boxen_plot_zz.svg')

# style of the plot
sns.set_style('darkgrid') # whitegrid, darkgrid, white, dark, ticks

# method 2 and we can change fonts size and fonts types 
sns.set_theme(style='dark', font_scale=1, font='Times New Roman') # whitegrid, darkgrid, white, dark, ticks

# change the legend position and title
plt.legend(loc='upper right', title='Class Name')

# change the x tick labels
plt.xticks(ticks=[0,1], labels=['Male', 'Female'], rotation=60)

# Rotat x tick label by 45 degree
#plt.xticks(rotation=45)
```

# Bar plot using seaborn

```python
# Barplot using seaborn

# st=le of the plot
sns.set_theme(style='darkgrid')

# Bar plot size of figure size
plt.figure(figsize=(12,6))
# change the scatter point colors 
pallete = {'First': 'green', 'Second': 'red', 'Third': 'blue'} # pallete mean color scheme / color mapping / color counter
sns.barplot(data= df, x= 'sex', y= 'age', hue='class', palette=pallete)

# laabels and title

# labels 
xlab = 'Gender'
ylab = 'Age'
plt.xlabel(xlab)
plt.ylabel(ylab)

# title
plt.title('Bar Plot Of Age (Based On Gender) And Class Name Of Passengers')

# if we save to the file in other folder 

# saving a plot 300dpi, 600dpi into png file and size of the plot 
plt.savefig('./outputs/plots/Bar_plot_zz.png', dpi=300)

# saving a plot in vector file such as svg file
plt.savefig('./outputs/plots/Bar_plot_zz.svg')

# style of the plot
sns.set_style('darkgrid') # whitegrid, darkgrid, white, dark, ticks

# method 2 and we can change fonts size and fonts types 
sns.set_theme(style='dark', font_scale=1, font='Times New Roman') # whitegrid, darkgrid, white, dark, ticks

# change the legend position and title
plt.legend(loc='upper right', title='Class Name')

# change the x tick labels
plt.xticks(ticks=[0,1], labels=['Male', 'Female'], rotation=60)

# Rotat x tick label by 45 degree
#plt.xticks(rotation=45)
```
