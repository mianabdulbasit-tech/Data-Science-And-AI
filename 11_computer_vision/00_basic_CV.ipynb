# Image Transformations And Manipulation Techniques

## 1. Resizing and Scaling Images

```python
image = cv2.imread("./images/python_image.png")

if image is None:
    print("Image not found")
else:
    print("Image found")
    resized = cv2.resize(image, (300, 300))

    cv2.imshow("Originla Image is ", image)
    cv2.imshow("resize image is", resized)

    cv2.imwrite("./images/resized_output.png", resized)

    cv2.waitKey(0)
    cv2.destroyAllWindows()

```

## 2. Cropping Images Using Slicing in Opencv

```python
image = cv2.imread("./images/python_image.png")

if image is not None:
    cropped = image[100:200, 50:150]
    cv2.imshow("Originla Image is ", image)
    cv2.imshow("cropped Image", cropped)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Image not found")
```

## 3.Image Rotation And flipping

```python
# rotating the image
image = cv2.imread("./images/python_image.png")

if image is not None:
    (h, w) = image.shape[:2]
    center = (w//2, h//2)

    M = cv2.getRotationMatrix2D(center, 45, 1.0)
    rotated_image = cv2.warpAffine(image, M, (w, h))

    cv2.imshow("Originla Image is ", image)
    cv2.imshow("rotated Image", rotated_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Image not found")
```

```python
# rotating the image other way
image = cv2.imread("./images/python_image.png")

if image is not None:
    rotated = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    cv2.imshow("Originla Image is ", image)
    cv2.imshow("rotated Image", rotated)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Image not found")
```

```python
# flipping the image

image = cv2.imread("./images/python_image.png")

if image is not None:
    flipped_h = cv2.flip(image, 1) # flip horizantally
    flipped_v = cv2.flip(image, 0) # vertically
    flipped_both = cv2.flip(image, -1) # both h and v

    cv2.imshow("Originla Image is ", image)
    cv2.imshow("flipped Image horizantal", flipped_h)
    cv2.imshow("flipped Image  vertical", flipped_v)
    cv2.imshow("flipped Image in both H And V", flipped_both)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Image not found")
```

# Basic Image Drawing Techniques

## 1. Drawing a Line-cv2.line()

```python
# draw a line 
image = cv2.imread("./images/python_image.png")

if image is not None:
    print("image loaded successfully")

    pt1 = (50, 100)
    pt2 = (300, 100)
    color = (255, 0, 0)
    thickness = 4

    cv2.line(image, pt1, pt2, color, thickness)

    cv2.imshow("Line Drawing", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Images not found")
```

## 2. Drawing a rectangle - cv2.rectangle()

```python
# draw a line 
import cv2

image = cv2.imread("./images/python_image.png")

if image is not None:
    print("image loaded successfully")

    pt1 = (50, 50)
    pt2 = (400, 350)
    color = (0, 0, 255)
    thickness = 3

    cv2.rectangle(image, pt1, pt2, color, thickness)

    cv2.imshow("Image focusing Rectangle", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Images not found")
```

## 3.Draw Circle - cv2.circle()

```python
# draw a line 
import cv2

image = cv2.imread("./images/python_image.png")

if image is not None:
    print("image loaded successfully")

    cv2.circle(image, (400, 200), 100, (255, 0, 0), 5)

    cv2.imshow("Image focusing Circle", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Images not found")
```

## 4.Adding text - cv2.putText()

```python
# draw a line 
import cv2

image = cv2.imread("./images/python_image.png")

if image is not None:
    print("image loaded successfully")
    #text = "Mian Abdul Basit"
    cv2.putText(image, "Mian Abdul Basit",  (250, 500),
                 cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 0, 255), 6)

    cv2.imshow("Adding Text over Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
else:
    print("Images not found")
```

```python
import cv2

# Step 1: Take file input from user
file_path = input("Enter image file path: ")
img = cv2.imread("./images/python_image.png")

if img is None:
    print("Error: Image not found!")
    exit()

# Step 2: Ask user what to draw
print("Choose an option to modify the image:")
print("1. Add Text")
print("2. Draw Line")
print("3. Draw Circle")
print("4. Draw Rectangle")

choice = input("Enter your choice (1-4): ")

# Step 3: Switch-case using match-case (Python 3.10+)
match choice:
    case "1":
        cv2.putText(img, "Hello OpenCV", (50, 50),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        print("Text added.")

    case "2":
        cv2.line(img, (50, 50), (300, 50), (255, 0, 0), 3)
        print("Line drawn.")

    case "3":
        cv2.circle(img, (150, 150), 50, (0, 0, 255), 3)
        print("Circle drawn.")

    case "4":
        cv2.rectangle(img, (50, 50), (200, 200), (0, 255, 255), 3)
        print("Rectangle drawn.")

    case _:
        print("Invalid choice!")
        exit()

# Step 4: Show and save result
cv2.imshow("Modified Image", img)
cv2.imwrite("output.jpg", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

def add_text(img):
    cv2.putText(img, "Hello OpenCV", (50, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

def draw_line(img):
    cv2.line(img, (50, 50), (300, 50), (255, 0, 0), 3)

def draw_circle(img):
    cv2.circle(img, (150, 150), 50, (0, 0, 255), 3)

def draw_rectangle(img):
    cv2.rectangle(img, (50, 50), (200, 200), (0, 255, 255), 3)

# Simulated switch-case using dictionary
switch_case = {
    "1": add_text,
    "2": draw_line,
    "3": draw_circle,
    "4": draw_rectangle
}

file_path = input("Enter image file path: ")
img = cv2.imread("./images/python_image.png")

if img is None:
    print("Error: Image not found!")
    exit()

print(" 1. Text ")
print(" 2. Line ")
print(" 3. Circle ")
print(" 4. Rectangle")

choice = input("Enter your choice (1-4): ")

func = switch_case.get(choice)
if func:
    func(img)
    cv2.imshow("Modified Image", img)
    cv2.imwrite("output.jpg", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Invalid choice!")
```
